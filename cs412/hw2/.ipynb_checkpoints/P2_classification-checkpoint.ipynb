### 1. Choice of Dataset
Slate, D. (1991). Letter Recognition [Dataset]. UCI Machine Learning Repository. https://doi.org/10.24432/C5ZP40.

This is a dataset of character image features with the goal to identify the letter shown.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from ucimlrepo import fetch_ucirepo 
from sklearn.model_selection import KFold
  
# fetch dataset 
letter_recognition = fetch_ucirepo(id=59) 
  
# data (as pandas dataframes) 
X = letter_recognition.data.features 
y = letter_recognition.data.targets 
  
# metadata 
print(letter_recognition.metadata) 
  
# variable information 
print(letter_recognition.variables) 
```

### 2. Implement Logistic Regression

Binary target: The letter A versus all other letters, i.e. 1 if letter == 'A', else 0

```python
X_all = X.to_numpy(dtype=np.float64)
y_labels = y.iloc[:, 0].astype(str).to_numpy()
y_all = (y_labels == "A").astype(np.float64)

rng = np.random.default_rng(12)
idx = rng.choice(X_all.shape[0], size=5000, replace=False)

X_sub = X_all[idx]
y_sub = y_all[idx]

X_train, X_test, y_train, y_test = train_test_split(
    X_sub, y_sub,
    test_size=0.4,
    random_state=12,
    stratify=y_sub
)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test  = scaler.transform(X_test)

X_train_b = np.c_[np.ones((X_train.shape[0], 1)), X_train]
X_test_b  = np.c_[np.ones((X_test.shape[0], 1)),  X_test]

def sigmoid(z):
    z = np.clip(z, -500, 500)
    return 1.0 / (1.0 + np.exp(-z))

def fit_logreg_gd(X, y, lr=0.1, epochs=2000, reg_type=None, lam=0.0):
    n, d = X.shape
    w = np.zeros(d)

    for _ in range(epochs):
        p = sigmoid(X @ w)
        grad = (X.T @ (p - y)) / n

        if reg_type == "l2":
            grad[1:] += lam * w[1:]

        elif reg_type == "l1":
            grad[1:] += lam * np.sign(w[1:])

        w -= lr * grad

    return w

w = fit_logreg_gd(X_train_b, y_train, lr=0.1, epochs=2000)

p_test = sigmoid(X_test_b @ w)

squared_loss = np.mean((p_test - y_test) ** 2)
absolute_loss = np.mean(np.abs(p_test - y_test))

print("Test Squared Loss:", squared_loss)
print("Test Absolute Loss:", absolute_loss)

y_pred = (p_test >= 0.5).astype(int)
accuracy = np.mean(y_pred == y_test)
print("Test Accuracy:", accuracy)
```

### 3. Dataset Scaling and Reporting Observations

```python
def cross_validate(X, y, reg_type, lambdas, k=5, lr=0.1, epochs_cv=800):
    kf = KFold(n_splits=k, shuffle=True, random_state=12)
    best_lambda, best_loss = None, float("inf")

    for lam in lambdas:
        fold_losses = []
        for tr_idx, val_idx in kf.split(X):
            X_tr, X_val = X[tr_idx], X[val_idx]
            y_tr, y_val = y[tr_idx], y[val_idx]

            w = fit_logreg_gd(X_tr, y_tr, lr=lr, epochs=epochs_cv, reg_type=reg_type, lam=lam)
            p_val = sigmoid(X_val @ w)
            fold_losses.append(np.mean((p_val - y_val) ** 2))

        avg_loss = np.mean(fold_losses)
        if avg_loss < best_loss:
            best_loss, best_lambda = avg_loss, lam

    return best_lambda

def run_experiment_for_N(X_all, y_all, N, train_sizes, lambdas, k=5, seed=12, lr=0.1, epochs_final=2000, epochs_cv=800):
    rng = np.random.default_rng(seed)
    idx = rng.choice(X_all.shape[0], size=N, replace=False)
    X_sub = X_all[idx]
    y_sub = y_all[idx]

    print(f"\nDataset size N={N}")
    print("Positive rate overall:", y_all.mean())
    print("Positive rate in sampled subset:", y_sub.mean())

    results = {
        "N": N,
        "train_sizes": train_sizes,
        "no_reg": [],
        "l1": [],
        "l2": [],
        "best_l1": [],
        "best_l2": []
    }

    for size in train_sizes:
        X_train, X_test, y_train, y_test = train_test_split(
            X_sub, y_sub, train_size=size, random_state=12, stratify=y_sub
        )

        scaler = StandardScaler()
        X_train = scaler.fit_transform(X_train)
        X_test  = scaler.transform(X_test)

        X_train = np.c_[np.ones((X_train.shape[0], 1)), X_train]
        X_test  = np.c_[np.ones((X_test.shape[0], 1)),  X_test]

        # No regularization
        w = fit_logreg_gd(X_train, y_train, lr=lr, epochs=epochs_final)
        p = sigmoid(X_test @ w)
        results["no_reg"].append(np.mean((p - y_test) ** 2))

        # L1 regularization
        best_l1 = cross_validate(X_train, y_train, "l1", lambdas, k=k, lr=lr, epochs_cv=epochs_cv)
        w = fit_logreg_gd(X_train, y_train, lr=lr, epochs=epochs_final, reg_type="l1", lam=best_l1)
        p = sigmoid(X_test @ w)
        results["l1"].append(np.mean((p - y_test) ** 2))
        results["best_l1"].append(best_l1)
        
        # L2 regularization
        best_l2 = cross_validate(X_train, y_train, "l2", lambdas, k=k, lr=lr, epochs_cv=epochs_cv)
        w = fit_logreg_gd(X_train, y_train, lr=lr, epochs=epochs_final, reg_type="l2", lam=best_l2)
        p = sigmoid(X_test @ w)
        results["l2"].append(np.mean((p - y_test) ** 2))
        results["best_l2"].append(best_l2)

    return results


train_sizes = [0.2, 0.4, 0.6, 0.8]
lambdas = [1e-4, 1e-3, 1e-2, 1e-1]

all_results = []
for N in [5000, 10000, 20000]:
    res = run_experiment_for_N(X_all, y_all, N, train_sizes, lambdas, k=5, seed=12, lr=0.1, epochs_final=2000, epochs_cv=800)
    all_results.append(res)

for res in all_results:
    plt.figure()
    plt.plot(res["train_sizes"], res["no_reg"], label="No Reg")
    plt.plot(res["train_sizes"], res["l1"], label="L1")
    plt.plot(res["train_sizes"], res["l2"], label="L2")
    plt.xlabel("Training Size")
    plt.ylabel("Test Squared Loss")
    plt.title(f"Test Loss vs Training Size (N={res['N']})")
    plt.legend()
    plt.show()
```

#### Observations:

Across all settings, test squared loss is low (roughly in the $10^{-3}$ to $10^{-2}$ range), indicating the binary task (A vs not-A) is learnable with this feature set. In all three dataset sizes ($N=5000, 10000, 20000$), performance improves from small training fractions and reaches its best region, which is around 40%–60% training set size. At 80% training set size, test loss increases. Since test size becomes smaller, this estimate has higher variance and this increase is likely due to split variability rather than model failure. L1, L2, and no-regularization curves are very close, meaning regularization has limited marginal impact. When there is a difference, L2 is usually slightly better than L1, while no regularization is often the best option.


Moving from smaller to larger total datasets keeps the same general trend (best performance around 40%-60% training set size), but the curves become more stable. With larger datasets, the model maintains consistently low test loss, showing improved generalization from having more diverse training examples. Regularization still provides only a small change in loss and performs better than L1 and L2 regularization. Overall, increasing dataset size helps generalization reliability more than changing between L1 and L2 regularization in this specific case. For this specific problem of predicting the letter 'A' versus all other letters prioritizing  splitting the data at around a 40%-60% training set size is the most important for minimizing test squared loss.


### 4. Extend to Multiclass Logistic Regression

```python
X_all = X.to_numpy(dtype=np.float64)
y_labels = y.iloc[:, 0].astype(str).to_numpy()

classes = np.unique(y_labels)
class_to_int = {c:i for i,c in enumerate(classes)}
y_int_all = np.array([class_to_int[c] for c in y_labels], dtype=np.int64)

K = len(classes)

def one_hot(y, K):
    Y = np.zeros((y.shape[0], K), dtype=np.float64)
    Y[np.arange(y.shape[0]), y] = 1.0
    return Y

def softmax(Z):
    Z = Z - np.max(Z, axis=1, keepdims=True)
    expZ = np.exp(np.clip(Z, -500, 500))
    return expZ / np.sum(expZ, axis=1, keepdims=True)

def fit_softmax_gd(X, y_int, K, lr=0.1, epochs=2000, reg_type=None, lam=0.0):
    n, d = X.shape
    W = np.zeros((d, K), dtype=np.float64)
    Y = one_hot(y_int, K)

    for _ in range(epochs):
        P = softmax(X @ W)
        grad = (X.T @ (P - Y)) / n

        if reg_type == "l2":
            grad[1:, :] += lam * W[1:, :]
        elif reg_type == "l1":
            grad[1:, :] += lam * np.sign(W[1:, :])

        W -= lr * grad

    return W

def multiclass_squared_loss(P, y_int, K):
    Y = one_hot(y_int, K)
    return np.mean(np.sum((P - Y)**2, axis=1))

def multiclass_absolute_loss(P, y_int, K):
    Y = one_hot(y_int, K)
    return np.mean(np.sum(np.abs(P - Y), axis=1))

def multiclass_accuracy(P, y_int):
    return np.mean(np.argmax(P, axis=1) == y_int)

def cross_validate_softmax(X, y_int, K, reg_type, lambdas, k=5, lr=0.1, epochs_cv=800):
    kf = KFold(n_splits=k, shuffle=True, random_state=12)
    best_lam, best_loss = None, float("inf")

    for lam in lambdas:
        fold_losses = []

        for tr_idx, val_idx in kf.split(X):
            X_tr, X_val = X[tr_idx], X[val_idx]
            y_tr, y_val = y_int[tr_idx], y_int[val_idx]

            W = fit_softmax_gd(X_tr, y_tr, K, lr=lr, epochs=epochs_cv, reg_type=reg_type, lam=lam)
            P_val = softmax(X_val @ W)

            loss = multiclass_squared_loss(P_val, y_val, K)
            fold_losses.append(loss)

        avg_loss = np.mean(fold_losses)
        if avg_loss < best_loss:
            best_loss, best_lam = avg_loss, lam

    return best_lam

def run_experiment_for_N_multiclass(X_all, y_int_all, K, N, train_sizes, lambdas, k=5, seed=12, lr=0.1, epochs_final=2000, epochs_cv=800):
    rng = np.random.default_rng(seed)
    idx = rng.choice(X_all.shape[0], size=N, replace=False)
    X_sub = X_all[idx]
    y_sub = y_int_all[idx]

    results = {
        "N": N,
        "train_sizes": train_sizes,
        "no_reg": [],
        "l1": [],
        "l2": [],
        "best_l1": [],
        "best_l2": [],
        "acc_no_reg": [],
        "acc_l1": [],
        "acc_l2": []
    }

    for size in train_sizes:
        X_train, X_test, y_train, y_test = train_test_split(
            X_sub, y_sub, train_size=size, random_state=12, stratify=y_sub
        )

        scaler = StandardScaler()
        X_train = scaler.fit_transform(X_train)
        X_test  = scaler.transform(X_test)

        X_train = np.c_[np.ones((X_train.shape[0], 1)), X_train]
        X_test  = np.c_[np.ones((X_test.shape[0], 1)),  X_test]

        # No regularization
        W = fit_softmax_gd(X_train, y_train, K, lr=lr, epochs=epochs_final)
        P = softmax(X_test @ W)
        results["no_reg"].append(multiclass_squared_loss(P, y_test, K))
        results["acc_no_reg"].append(multiclass_accuracy(P, y_test))

        # L1 regularization
        best_l1 = cross_validate_softmax(X_train, y_train, K, "l1", lambdas, k=k, lr=lr, epochs_cv=epochs_cv)
        W = fit_softmax_gd(X_train, y_train, K, lr=lr, epochs=epochs_final, reg_type="l1", lam=best_l1)
        P = softmax(X_test @ W)
        results["l1"].append(multiclass_squared_loss(P, y_test, K))
        results["best_l1"].append(best_l1)
        results["acc_l1"].append(multiclass_accuracy(P, y_test))

        # L2 regularization
        best_l2 = cross_validate_softmax(X_train, y_train, K, "l2", lambdas, k=k, lr=lr, epochs_cv=epochs_cv)
        W = fit_softmax_gd(X_train, y_train, K, lr=lr, epochs=epochs_final, reg_type="l2", lam=best_l2)
        P = softmax(X_test @ W)
        results["l2"].append(multiclass_squared_loss(P, y_test, K))
        results["best_l2"].append(best_l2)
        results["acc_l2"].append(multiclass_accuracy(P, y_test))

    return results

train_sizes = [0.2, 0.4, 0.6, 0.8]
lambdas = [1e-4, 1e-3, 1e-2, 1e-1]

all_results_mc = []
for N in [5000, 10000, 20000]:
    result = run_experiment_for_N_multiclass(X_all, y_int_all, K, N, train_sizes, lambdas, k=5, seed=12, lr=0.1, epochs_final=2000, epochs_cv=800)
    all_results_mc.append(result)

for res in all_results_mc:
    plt.figure()
    plt.plot(res["train_sizes"], res["no_reg"], label="No Reg")
    plt.plot(res["train_sizes"], res["l1"], label="L1")
    plt.plot(res["train_sizes"], res["l2"], label="L2")
    plt.xlabel("Training Size")
    plt.ylabel("Test Squared Loss (Multiclass)")
    plt.title(f"Multiclass Test Loss vs Training Size (N={res['N']})")
    plt.legend()
    plt.show()
```
